# Test Chat Service

This notebook tests the chat service functionality.


```python
import os
from dotenv import load_dotenv
from services.chat_service import ChatService

# Load environment variables
load_dotenv()

# Check if API key is set
api_key = os.getenv("OPENAI_API_KEY")
if not api_key:
    print("ERROR: OPENAI_API_KEY not found in environment variables")
    print("Please set it in your .env file or as an environment variable")
else:
    print(f"✓ OpenAI API key found (starts with: {api_key[:7]}...)")
```

```python
# Initialize chat service
try:
    chat_service = ChatService()
    print("✓ Chat service initialized successfully")
    print(f"Model: {chat_service.model}")
except Exception as e:
    print(f"✗ Error initializing chat service: {e}")
```

## Test 1: Send a message and stream response


```python
# Test user ID (you can use any integer)
test_user_id = 1

# Send a test message
test_message = "Hi, I'd like to start planning for my retirement. I'm 35 years old."

print(f"User ({test_user_id}): {test_message}")
print("\nAssistant:")
print("-" * 50)

# Stream the response
full_response = ""
for chunk in chat_service.send_message(test_user_id, test_message):
    print(chunk, end="", flush=True)  # Print chunks as they arrive
    full_response += chunk

print("\n" + "-" * 50)
```

## Test 2: Continue the conversation


```python
# Send a follow-up message
follow_up = "I currently have $50,000 in my retirement account."

print(f"User ({test_user_id}): {follow_up}")
print("\nAssistant:")
print("-" * 50)

for chunk in chat_service.send_message(test_user_id, follow_up):
    print(chunk, end="", flush=True)

print("\n" + "-" * 50)
```

## Test 3: View chat history


```python
# Get the full chat history
history = chat_service.get_chat_history(test_user_id)

print(f"Chat history for user {test_user_id}:")
print("=" * 50)
for i, msg in enumerate(history, 1):
    print(f"\n[{i}] {msg['role'].upper()}:")
    print(msg['content'])
    print("-" * 50)
```

## Test 4: Export chat history


```python
# Export chat history as text
exported_text = chat_service.export_chat_history(test_user_id)

print("Exported chat history:")
print("=" * 50)
print(exported_text)
print("=" * 50)

# You could save this to a file like the parser expects:
# with open('test_conversation.txt', 'w') as f:
#     f.write(exported_text)
```

## Test 5: Clear chat history


```python
# Clear the chat history
chat_service.clear_chat_history(test_user_id)

# Verify it's cleared
history_after_clear = chat_service.get_chat_history(test_user_id)
print(f"History after clearing: {len(history_after_clear)} messages")
print("✓ Chat history cleared successfully")
```
